```python
!pip install gTTS

from gtts import gTTS
from IPython.display import Audio, display
import os
import tempfile

print("Welcome to the Robo Speaker!")

while True:
    x = input("Enter what you want me to pronounce (or 'z' to exit): ")

    if x.lower() == 'z':
        x = "Bye Bye Friend"
        print(x)

        tts = gTTS(text=x, lang='en')
        with tempfile.NamedTemporaryFile(suffix=".mp3", delete=False) as f:
            filename = f.name

        tts.save(filename)
        display(Audio(filename, autoplay=True))
        os.remove(filename)
        break

    if x.strip():
        tts = gTTS(text=x, lang='en')

        with tempfile.NamedTemporaryFile(suffix=".mp3", delete=False) as f:
            filename = f.name

        tts.save(filename)
        display(Audio(filename, autoplay=True))
        os.remove(filename)
```
