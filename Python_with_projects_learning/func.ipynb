```python
# 1. Func without parameters and without return value
def greeting():
    print("Namaste")
greeting() 
```

```python
# 2.Function with parameters and no return value.
def add(a,b):
    print("sum =",a+b)
add(10,20)
add(5,7)
```

```python
# 3.Function with a return value.
def sq(n):
    return n*n
result=sq(5)
print("square=",result)
```

```python
# 4. Function with default parameters
def greet(name="Friend"):
    print("Hello,", name)

greet()
greet("Ram")
```

```python
# Exception Handling
try:
    a=int(input("Enter a number"))
    print(a+10)
except:
    print("Some Error occurred")
    
```
