```python
import pandas as pd
df=pd.read_csv('messy_data.csv')

df.isnull().any() # Search a missing value
# df.isnull().sum()     # No of missing value
# df.fillna(0)
# print("\n",df)
```

```python
# Filling a missing values with mean of the column.
df['new_age']=df['age'].fillna(df['age'].mean())
df['new_score']=df['score'].fillna(df['score'].mean())
df.dtypes
```

```python
# Renaming a columns
df=df.rename(columns={'email':'e-mail'})
df.head(10)
```

```python
# Change datatypes

df['new_age']=df['age'].fillna(df['age'].mean()).astype(int)
df.head()
```

```python
df['new_score']=df['score'].apply(lambda x:x+10)
df.head()
```

```python
# Merging and joining DataFrames
df1=pd.DataFrame({'key':['A','B','C'],'value1':[1,2,3]})
df2=pd.DataFrame({'key':['A','B','D'],'value2':[4,5,6]})
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1,df2,on="key",how='outer')
```

```python
pd.merge(df1,df2,on="key",how='inner')
```

```python
pd.merge(df1,df2,on="key",how='left')
```

```python
pd.merge(df1,df2,on="key",how='right')
```
