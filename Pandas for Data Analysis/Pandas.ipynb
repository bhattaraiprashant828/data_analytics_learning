```python
#1. Pandas Series.
import pandas as pd
data=[1,2,3,4,5]
series=pd.Series(data)
print("Series \n",series)
print(type(series))
```

```python
# 2.Create a series from dictonary
import pandas as pd
data={
    'a':1,
    'b':2,
    'c':3
    }
pd.Series(data)
```

```python
# 3. Create series
data=[10,20,30]
index=['a','b','c']
pd.Series(data,index)
```

```python
# Dataframe
data={
    'Name':['Krish','John','Jack'],
    'Age':[25,30,45],
    'city':['Banglore','Kathmandu','Pokhara']
    }
df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
# Create a dataframe from a list of dictonaries
data=[
        {'Name':'Krish','Age':32,'city':'Kathmandu'},
        {'Name':'John','Age':36,'city':'Pokhara'},
        {'Name':'Jack','Age':34,'city':'Kapilvastu'},
]
df=pd.DataFrame(data)
print(df)
print(type(df))

df['Salary']=[10000,200000,3000]  # Adds a column Salary
print("\n",df)

df.drop('Salary',axis=1,inplace=True) ## Remove a column Salary
print("\n",df)

df['Age']=df['Age']+1 ## Add a age to a column
print("\n",df)

df.drop(0,inplace=True) ## Remove a  Row 
print("\n",df)
```

```python
# # Accessing Data From DataFrame
print(df)
print("\n",df.loc[0])   ## loc access the row index.
print("\n",df.iloc[2])  ## iloc access the column index.
```
